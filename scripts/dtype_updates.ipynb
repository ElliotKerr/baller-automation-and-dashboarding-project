```python
import sqlite3
import pandas as pd
from statsbombpy import sb
from datetime import datetime
from sqlalchemy import String, Integer, DateTime, Float, JSON, Boolean, Date

pd.set_option('display.max_columns', None)
```

```python
competitions = sb.competitions()
```

```python
SQL_PY_DTYPE_MAPPING = {
    String: str,
    Integer: int,
    DateTime: datetime,
    Float: float,
    JSON: dict,
    Boolean: bool
}

column_mapping = {
    "competition_id": Integer,
    "season_id": Integer,
    "country_name": String,
    "competition_name": String,
    "competition_gender": String,
    "competition_youth": Boolean,
    "competition_international": Boolean,
    "season_name": String,
    "match_updated": DateTime,
    "match_updated_360": DateTime,
    "match_available_360": DateTime,
    "match_available": DateTime,
}
```

```python
def col_cleaning(df, column_mapping, dtype_mapping):
    """
    Doc String
    """
    for col, sql_dtype in column_mapping.items():
        py_type = dtype_mapping[sql_dtype]

        df[col] = df[col].apply(lambda x: None if x in ('NaN', 'None', 'Null') else x)

        if py_type is datetime:
            df[col] = pd.to_datetime(df[col], format="ISO8601", utc=True)
        if py_type is int:
            df[col] = pd.to_numeric(df[col]).apply(lambda x: int(x))
        if py_type is float:
            df[col] = pd.to_numeric(df[col]).apply(lambda x: float(x))

    return df
```

```python
df = col_cleaning(competitions, column_mapping, SQL_PY_DTYPE_MAPPING)
```
