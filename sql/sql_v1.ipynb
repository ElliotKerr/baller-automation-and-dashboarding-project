```python
import sqlite3
import pandas as pd

pd.set_option('display.max_columns', None)
```

```python
connection = sqlite3.connect(f'../dbs/gold.db')
cursor = connection.cursor()

cursor.execute("ATTACH DATABASE '../dbs/bronze.db' AS bronze")
cursor.execute("ATTACH DATABASE '../dbs/silver.db' AS silver")
```

```python
def load_sql_table(query, connection):
    df = pd.read_sql_query(query, connection)
    return df
```

```python
query = f"""
    SELECT
        *
    FROM
        silver.complete
    WHERE 
        LOWER(type) LIKE '%pass%'
    LIMIT 100
"""
```

```python
silver_complete = load_sql_table(query, connection)
```
