```python
import sqlite3
import pandas as pd

pd.set_option('display.max_columns', None)
```

```python
comp_match_query = f"""
    SELECT 
        c.country_name
        ,c.competition_name	
        ,c.competition_gender
        ,c.competition_youth
        ,c.competition_international
        ,c.season_name
        ,m.*
    FROM 
        competitions c
    LEFT JOIN 
        matches m
    ON 
        c.competition_id = m.competition_id
    AND 
        c.season_id = m.season_id
"""
```

```python
connection = sqlite3.connect('../dbs/bronze.db')

# 2. Read query directly into a DataFrame
df = pd.read_sql_query(comp_match_query, connection)

# 3. Always close the connection when finished
connection.close()

df
```

```python
import sqlite3

def create_bronze_table_structure():
    _dict = {
        'bronze':['competitions', 'matches', 'events']
    }
    for schema, table_list in _dict.items():
        with sqlite3.connect(f"../dbs/{schema}.db") as conn:
            cursor = conn.cursor()

            cursor.execute("ATTACH DATABASE '../dbs/staging.db' AS staging")

            try:
                for table in table_list:
                    cursor.execute("""
                        SELECT sql
                        FROM staging.sqlite_master
                        WHERE type = 'table'
                        AND name = ?
                    """, (table,))

                    row = cursor.fetchone()
                    if row is None:
                        print(f"{schema}: '{table}' not found in staging.db")
                        continue

                    create_sql = row[0]

                    cursor.execute(f'DROP TABLE IF EXISTS "{table}"')
                    cursor.execute(create_sql)

            finally:
                cursor.execute("DETACH DATABASE staging")
```

```python
create_bronze_table_structure()
```
