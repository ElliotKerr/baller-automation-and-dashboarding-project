```python
import sqlite3

# Connect to the database file
conn = sqlite3.connect('../dbs/competitions.db')
cursor = conn.cursor()

cursor.execute("""DROP TABLE IF EXISTS competitions""")
conn.commit()

# Get list of all tables
cursor.execute("""CREATE TABLE competitions AS 
               SELECT * FROM staging_competitions""")
conn.commit()

# Get list of all tables
cursor.execute("""DELETE FROM competitions""")
conn.commit()

conn.close()
```
